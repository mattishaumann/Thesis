```python
import pandas as pd

# Load topic assignments
df_assignments = pd.read_csv("01_topic_assignments.csv")

# Load topic groupings
df_groupings = pd.read_csv("02_topic_groupings.csv")
```

```python
df_assignments
```

```python
df_groupings
```

```python
# Join manual labels from df_groupings: topic_label_current (assignments) == topic_label (groupings)
manual_cols = ["topic_label", "Manual Topic Label", "Manual Cluster Label"]

_group = df_groupings[manual_cols].drop_duplicates(subset=["topic_label"], keep="first")

df_assignments = df_assignments.merge(
    _group,
    left_on="topic_label_current",
    right_on="topic_label",
    how="left",
).drop(columns=["topic_label"])

# Rows with no match keep NaNs in the manual columns
n_unmatched = df_assignments["Manual Topic Label"].isna().sum()
if n_unmatched:
    print(f"Warning: {n_unmatched} rows had no matching topic_label in df_groupings")
```
