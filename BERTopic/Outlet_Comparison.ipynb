# Similarity Matrix

```python
# Lean/reusable topic-embedding cosine matrix for any two outlet models

def _unwrap_model(m):
    return m["topic_model"] if isinstance(m, dict) and "topic_model" in m else m

def _top_topic_texts(topic_model, n_topics=25, top_k_words=15):
    info = topic_model.get_topic_info().copy()
    info = info[info["Topic"] != -1].nlargest(n_topics, "Count").reset_index(drop=True)

    topic_ids = info["Topic"].astype(int).tolist()
    topic_names = info["Name"].astype(str).tolist()
    topic_labels = [f"{tid} | {name}" for tid, name in zip(topic_ids, topic_names)]
    topic_texts = [
        " ".join([w for w, _ in (topic_model.get_topic(tid) or [])[:top_k_words]])
        for tid in topic_ids
    ]
    return info, topic_ids, topic_names, topic_labels, topic_texts

def cosine_topic_matrix(
    model_a,
    model_b,
    outlet_a="A",
    outlet_b="B",
    n_topics=25,
    top_k_words=15,
    encoder_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
):
    tm_a = _unwrap_model(model_a)
    tm_b = _unwrap_model(model_b)

    info_a, ids_a, names_a, labels_a, texts_a = _top_topic_texts(tm_a, n_topics, top_k_words)
    info_b, ids_b, names_b, labels_b, texts_b = _top_topic_texts(tm_b, n_topics, top_k_words)

    encoder = SentenceTransformer(encoder_name)
    emb_a = encoder.encode(texts_a, normalize_embeddings=True)
    emb_b = encoder.encode(texts_b, normalize_embeddings=True)

    sim = cosine_similarity(emb_a, emb_b)  # shape: (n_topics_a, n_topics_b)
    sim_df = pd.DataFrame(sim, index=labels_a, columns=labels_b)

    # Best match in B for each topic in A
    j_best = sim.argmax(axis=1)
    best_matches = pd.DataFrame({
        f"{outlet_a}_topic_id": ids_a,
        f"{outlet_a}_topic_name": names_a,
        f"{outlet_b}_topic_id": [ids_b[j] for j in j_best],
        f"{outlet_b}_topic_name": [names_b[j] for j in j_best],
        "cosine_similarity": [float(sim[i, j]) for i, j in enumerate(j_best)],
    }).sort_values("cosine_similarity", ascending=False).reset_index(drop=True)

    return sim_df, best_matches

# ---- Example usage (adapt names to your notebook) ----
models = {
    "tagesschau": tm_ts,
    "rt": tm_rt,
    "compact": tm_cm,
    "nius": tm_ns,
    "tichys": tm_te,
    "antispiegel": tm_as,
    "dk": tm_dk,
}

```

## Weighted best-match outlet similarity for BERTopic models

```python
# Computes directional and symmetric outlet similarity scores based on topic embeddings.
# S(A,B) = sum_t [ w_t * max_u cosine(emb_t, emb_u) ]
# where w_t are prevalence weights from topic counts in outlet A.

from __future__ import annotations

from typing import Dict, Tuple, Optional, List



def _select_topics(
    topic_info: pd.DataFrame,
    min_topic_size: int = 1,
    top_n_topics: Optional[int] = None,
) -> pd.DataFrame:
    """
    Select valid topics from BERTopic topic_info:
    - exclude topic -1
    - apply minimum topic count
    - optionally keep top_n by count
    """
    if not {"Topic", "Count"}.issubset(topic_info.columns):
        raise ValueError("topic_info must contain columns: 'Topic', 'Count'.")

    df = topic_info.copy()
    df = df[df["Topic"] != -1].copy()
    df["Topic"] = df["Topic"].astype(int)
    df["Count"] = pd.to_numeric(df["Count"], errors="coerce").fillna(0).astype(int)
    df = df[df["Count"] >= int(min_topic_size)].copy()

    if top_n_topics is not None:
        df = df.nlargest(int(top_n_topics), "Count")

    # stable order for reproducibility
    df = df.sort_values("Topic").reset_index(drop=True)
    return df


def _topic_embedding_map(model) -> Dict[int, np.ndarray]:
    """
    Build mapping: topic_id -> embedding vector from a BERTopic model.
    Handles common BERTopic embedding indexing variants.
    """
    if not hasattr(model, "topic_embeddings_"):
        raise ValueError("Model has no `topic_embeddings_` attribute.")

    emb = np.asarray(model.topic_embeddings_)
    if emb.ndim != 2:
        raise ValueError("`topic_embeddings_` must be a 2D array.")

    topics_all = sorted(int(t) for t in model.get_topics().keys())
    topics_no_outlier = [t for t in topics_all if t != -1]

    mapping: Dict[int, np.ndarray] = {}

    # Case 1: embeddings include -1 and all topic ids
    if emb.shape[0] == len(topics_all):
        mapping = {tid: emb[i] for i, tid in enumerate(topics_all)}
    # Case 2: embeddings exclude -1
    elif emb.shape[0] == len(topics_no_outlier):
        mapping = {tid: emb[i] for i, tid in enumerate(topics_no_outlier)}
    else:
        # Fallback: try topic_info ordering
        info_topics = model.get_topic_info()["Topic"].astype(int).tolist()
        info_no_outlier = [t for t in info_topics if t != -1]

        if emb.shape[0] == len(info_topics):
            mapping = {tid: emb[i] for i, tid in enumerate(info_topics)}
        elif emb.shape[0] == len(info_no_outlier):
            mapping = {tid: emb[i] for i, tid in enumerate(info_no_outlier)}
        else:
            raise ValueError(
                f"Cannot align embeddings (rows={emb.shape[0]}) with model topic ids."
            )

    # exclude outlier topic if present
    mapping.pop(-1, None)
    return mapping


def _pairwise_topic_matrix(
    topic_ids_a: List[int],
    emb_map_a: Dict[int, np.ndarray],
    topic_ids_b: List[int],
    emb_map_b: Dict[int, np.ndarray],
) -> pd.DataFrame:
    """
    Full topic-topic cosine matrix for outlet A vs outlet B.
    """
    if len(topic_ids_a) == 0 or len(topic_ids_b) == 0:
        return pd.DataFrame(index=topic_ids_a, columns=topic_ids_b, dtype=float)

    Xa = np.vstack([emb_map_a[t] for t in topic_ids_a])
    Xb = np.vstack([emb_map_b[t] for t in topic_ids_b])
    sim = cosine_similarity(Xa, Xb)
    return pd.DataFrame(sim, index=topic_ids_a, columns=topic_ids_b)


def compute_weighted_best_match_scores(
    models: Dict[str, object],
    topic_info_dict: Dict[str, pd.DataFrame],
    min_topic_size: int = 1,
    top_n_topics: Optional[int] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame, Dict[Tuple[str, str], pd.DataFrame]]:
    """
    Returns:
      - directional_scores: DataFrame with S(A,B)
      - symmetric_scores: DataFrame with S_sym(A,B) = (S(A,B)+S(B,A))/2
      - topic_topic_matrices: dict[(A,B)] -> full cosine topic-topic DataFrame
    """
    outlets = list(models.keys())
    if set(outlets) != set(topic_info_dict.keys()):
        raise ValueError("`models` and `topic_info_dict` must have identical keys.")

    # Precompute selected topics and embedding maps
    selected_topics = {}
    emb_maps = {}
    for outlet in outlets:
        selected_topics[outlet] = _select_topics(
            topic_info_dict[outlet],
            min_topic_size=min_topic_size,
            top_n_topics=top_n_topics,
        )
        emb_maps[outlet] = _topic_embedding_map(models[outlet])

        # ensure selected topics exist in embedding map
        missing = set(selected_topics[outlet]["Topic"]) - set(emb_maps[outlet].keys())
        if missing:
            raise ValueError(
                f"Outlet '{outlet}' has selected topics missing embeddings: {sorted(missing)[:10]}"
            )

    directional = pd.DataFrame(index=outlets, columns=outlets, dtype=float)
    matrices: Dict[Tuple[str, str], pd.DataFrame] = {}

    for a in outlets:
        topics_a = selected_topics[a]
        ids_a = topics_a["Topic"].tolist()
        counts_a = topics_a["Count"].to_numpy(dtype=float)

        if counts_a.sum() == 0 or len(ids_a) == 0:
            # no valid topics left after filtering
            directional.loc[a, :] = np.nan
            continue

        weights_a = counts_a / counts_a.sum()

        for b in outlets:
            ids_b = selected_topics[b]["Topic"].tolist()
            mat_ab = _pairwise_topic_matrix(ids_a, emb_maps[a], ids_b, emb_maps[b])
            matrices[(a, b)] = mat_ab

            if mat_ab.empty:
                directional.loc[a, b] = np.nan
                continue

            row_max = mat_ab.to_numpy().max(axis=1)  # best topic in B for each topic in A
            directional.loc[a, b] = float(np.dot(weights_a, row_max))

    symmetric = (directional + directional.T) / 2.0
    return directional, symmetric, matrices


def top_k_topic_matches(
    outlet_a: str,
    outlet_b: str,
    topic_topic_matrices: Dict[Tuple[str, str], pd.DataFrame],
    topic_info_dict: Dict[str, pd.DataFrame],
    k: int = 10,
) -> pd.DataFrame:
    """
    Return top-k topic pairs by cosine similarity for outlet A vs B.
    """
    mat = topic_topic_matrices[(outlet_a, outlet_b)]
    if mat.empty:
        return pd.DataFrame(
            columns=[
                "outlet_a",
                "topic_a",
                "topic_a_name",
                "outlet_b",
                "topic_b",
                "topic_b_name",
                "cosine_similarity",
            ]
        )

    pairs = (
        mat.stack()
        .reset_index()
        .rename(columns={"level_0": "topic_a", "level_1": "topic_b", 0: "cosine_similarity"})
        .sort_values("cosine_similarity", ascending=False)
        .head(k)
        .reset_index(drop=True)
    )

    info_a = topic_info_dict[outlet_a].copy()
    info_b = topic_info_dict[outlet_b].copy()
    name_map_a = dict(zip(info_a["Topic"].astype(int), info_a.get("Name", pd.Series([""] * len(info_a)))))
    name_map_b = dict(zip(info_b["Topic"].astype(int), info_b.get("Name", pd.Series([""] * len(info_b)))))

    pairs["outlet_a"] = outlet_a
    pairs["outlet_b"] = outlet_b
    pairs["topic_a_name"] = pairs["topic_a"].map(name_map_a)
    pairs["topic_b_name"] = pairs["topic_b"].map(name_map_b)

    return pairs[
        ["outlet_a", "topic_a", "topic_a_name", "outlet_b", "topic_b", "topic_b_name", "cosine_similarity"]
    ]

```

```python
models = {
    "tagesschau": tm_ts,
    "rt_de": tm_rt,
    "compact": tm_cm,
    "nius": tm_ns,
    "tichys": tm_te,
    "antispiegel": tm_as,
    "dk": tm_dk,
}

topic_info_dict = {name: model.get_topic_info() for name, model in models.items()}

directional_df, symmetric_df, topic_matrices = compute_weighted_best_match_scores(
    models=models,
    topic_info_dict=topic_info_dict,
    min_topic_size=10,      # optional
    top_n_topics=25,        # optional
)

# display(directional_df.round(4))
display(symmetric_df.round(4))

top_pairs_rt_ts = top_k_topic_matches(
    "rt_de", "tagesschau",
    topic_topic_matrices=topic_matrices,
    topic_info_dict=topic_info_dict,
    k=25
)
display(top_pairs_rt_ts)
```

## Tagesschau vs. RT

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from scipy.optimize import linear_sum_assignment

# --- Inputs ---
threshold = 0.70
top_k_words = 15  # words per topic for embedding text

# Use existing notebook variables if result_* dicts are missing
if "result_rt" in globals() and isinstance(result_rt, dict) and "topic_model" in result_rt:
    tm_rt = result_rt["topic_model"]

if "result_ts" not in globals():
    result_ts = {"topic_model": tm_ts}
tm_ts = result_ts["topic_model"]

# --- 1) Get ALL topics (exclude outlier -1) ---
rt_info = tm_rt.get_topic_info()
ts_info = tm_ts.get_topic_info()

rt_ids = rt_info.loc[rt_info["Topic"] != -1, "Topic"].astype(int).tolist()
ts_ids = ts_info.loc[ts_info["Topic"] != -1, "Topic"].astype(int).tolist()

# --- 2) Build topic texts ---
rt_texts = [" ".join([w for w, _ in (tm_rt.get_topic(t) or [])[:top_k_words]]) for t in rt_ids]
ts_texts = [" ".join([w for w, _ in (tm_ts.get_topic(t) or [])[:top_k_words]]) for t in ts_ids]

# --- 3) Encode + cosine matrix ---
encoder = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
rt_emb = encoder.encode(rt_texts, normalize_embeddings=True)
ts_emb = encoder.encode(ts_texts, normalize_embeddings=True)

sim_matrix = cosine_similarity(rt_emb, ts_emb)  # shape: (n_rt, n_ts)

# --- 4) One-to-one matching + threshold (prevents double-counting overlap) ---
r_idx, c_idx = linear_sum_assignment(-sim_matrix)  # maximize total similarity
matched_scores = sim_matrix[r_idx, c_idx]
keep = matched_scores >= threshold

common_rt = [rt_ids[i] for i in r_idx[keep]]
common_ts = [ts_ids[j] for j in c_idx[keep]]
common_scores = matched_scores[keep]

n_rt = len(rt_ids)
n_ts = len(ts_ids)
n_common = len(common_rt)
n_rt_only = n_rt - n_common
n_ts_only = n_ts - n_common

# Optional: table of common pairs
common_pairs = pd.DataFrame({
    "RT Topic": common_rt,
    "Tagesschau Topic": common_ts,
    "Cosine Similarity": common_scores
}).sort_values("Cosine Similarity", ascending=False)
display(common_pairs.head(20))

print(f"RT topics: {n_rt}")
print(f"Tagesschau topics: {n_ts}")
print(f"Common topics (cosine >= {threshold}): {n_common}")

# --- 5) Bubble overlap plot ---
fig, ax = plt.subplots(figsize=(8, 5))

left_center = (-0.6, 0)
right_center = (0.6, 0)
radius = 1.0

left_circle = Circle(left_center, radius, color="#f4b6c2", alpha=0.7, lw=0)
right_circle = Circle(right_center, radius, color="#bdbdbd", alpha=0.8, lw=0)
ax.add_patch(left_circle)
ax.add_patch(right_circle)

# counts
ax.text(-1.05, 0, str(n_rt_only), ha="center", va="center", fontsize=12)
ax.text(0.00, 0, str(n_common), ha="center", va="center", fontsize=12, fontweight="bold")
ax.text(1.05, 0, str(n_ts_only), ha="center", va="center", fontsize=12)

# labels
ax.text(-1.2, -1.25, "RT", ha="center", va="center", fontsize=12)
ax.text(1.2, -1.25, "Tagesschau", ha="center", va="center", fontsize=12)

ax.set_title(f"Topic Overlap (cosine ≥ {threshold})", fontsize=13)
ax.set_xlim(-2.2, 2.2)
ax.set_ylim(-1.6, 1.4)
ax.set_aspect("equal")
ax.axis("off")
plt.show()
```
