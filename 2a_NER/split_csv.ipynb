```python
import pandas as pd
import os
import math

INPUT_FILE = "/Users/MattisHaumann/Dev/Thesis/2a_NER/df_combined.csv"
OUTPUT_DIR = "/Users/MattisHaumann/Desktop/df_combined_splits"
MAX_MB = 30

os.makedirs(OUTPUT_DIR, exist_ok=True)

file_size_mb = os.path.getsize(INPUT_FILE) / (1024 ** 2)
print(f"Input file size: {file_size_mb:.1f} MB")

df = pd.read_csv(INPUT_FILE)
print(f"Total rows: {len(df):,}")

n_chunks = math.ceil(file_size_mb / MAX_MB)
print(f"Splitting into {n_chunks} chunks")

rows_per_chunk = math.ceil(len(df) / n_chunks)

for i in range(n_chunks):
    chunk = df.iloc[i * rows_per_chunk : (i + 1) * rows_per_chunk]
    out_path = os.path.join(OUTPUT_DIR, f"df_combined_part{i+1:02d}.csv")
    chunk.to_csv(out_path, index=False)
    size_mb = os.path.getsize(out_path) / (1024 ** 2)
    print(f"  Part {i+1}: {len(chunk):,} rows, {size_mb:.1f} MB → {out_path}")

print("Done.")
```
